```python
import numpy as np
```

```python
# int
arr = np.array([10,20,30,40])
print(arr,arr.dtype)
```

```python
# float
arr1 = np.array([10.1,10.2,10.3])
print(arr1, arr1.dtype)
```

```python
 # boolean
arr2 = np.array([True,False,True])
print(arr2, arr2.dtype)
```

```python
# complex_number
arr3 = np.array([10+1j,20+20j, 30+22j,40 +30j])
print(arr3, arr3.dtype)
```

```python
# strign 
arr4 = np.array(["kalpesh","aditya"])
print(arr4, arr4.dtype)
```

```python
# object

arr5 = np.array([[10, 20], "kalpesh", {"name": "kalpesh"}], dtype=object)
print(arr5, arr5.dtype)
```
