```python
import numpy as np
import time
```

```python
# checking execution time between python list and numpy
size = 10_000_000

python_list = list(range(size))
start_time = time.time()
sq_python_list = [x**2 for x in python_list]
end_time = time.time()
print(f"the actual time: {end_time - start_time} seconds")
```

```python
np_array = np.array(python_list)
start_time = time.time()
sq_np_array = np_array ** 2
end_time = time.time()
print(f"the actual time: {end_time - start_time} seconds")
```

```python
# check the size of store data
import sys
print(f"python list size: {sys.getsizeof(python_list) * len(python_list)}bytes")
print(f"numpy array size: {np_array.nbytes} bytes")
```
