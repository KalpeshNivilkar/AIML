```python
import numpy as np
```

```python
# reshape array
arr = np.array([[1,2,3],[4,5,6]])
print(arr, arr.shape)

reshape = arr.reshape((3,2))
print(reshape, reshape.shape)
```

```python
# flattened array meand 2d array converted into 1d array
flattene_arr = arr.flatten()
print(flattene_arr, flattene_arr.shape)
```

```python
# indexing

arr1 = np.array([1,2,3,4,5,6])
print(arr1[0])
```

```python
arr2 = np.array([[1,2,3],[4,5,6]])
print(arr2[1][1])
```

```python
# fancy_indexing
arr3 = np.array([1,2,3,4,5,10,7])
idx = [0,1,5]
print(arr3[idx])
```

```python
# boolean indexing
arr4 = np.array([10,20,30,40])
print(arr4[arr4 >= 20])
```

```python
# slicing
arr5 = np.array([10,20,30,40,50])
print(arr5[:])
print(arr5[1:])
print(arr5[:5])
print(arr5[::2])
```
