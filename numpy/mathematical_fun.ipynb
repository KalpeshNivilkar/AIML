```python
import numpy as np
```

```python
arr = np.array([1, 2, 3, 4, 5])  
print(np.sum(arr))  
print(np.prod(arr)) 
print(np.min(arr))  
# 15 
# 120  
# 1  
print(np.argmin(arr)) # 0 
print(np.max(arr))  
# 5  
print(np.argmax(arr)) # 4  
print(np.mean(arr))  
# 3.0  
print(np.median(arr))  # 3.0  
print(np.std(arr))  
# 1.41  
print(np.var(arr))
```

```python
# power function
arr = np.array([1, 2, 3, 4, 5])  
print(np.square(arr))  # [1, 4, 9, 16, 25] 
print(np.sqrt(arr))  
# [1, 1.41, 1.73, 2, 2.23]  
print(np.pow(arr, 4))  # [1, 8, 27, 64, 125] 
```

```python
# log 
print(np.log(arr))  
print(np.log10(arr))  
print(np.log2(arr))  
print(np.exp(arr)) 
```

```python
# rounding 
print(np.round(2.678)) # 3.0  
print(np.floor(2.678)) # 2.0 
print(np.ceil(2.678))  # 3.0  
print(np.trunc(2.678)) # 2.0
```

```python
# Miscellaneous Functions 
arr = np.array([1, 2, -5, 3, 8, -4, 2, 5]) 
print(np.abs(arr))  
print(np.sort(arr))  
# [1 2 5 3 8 4 2 5]  
# [-5 -4 1 2 2 3 5 8] 
print(np.unique(arr))  # [-5 -4 1 2 3 5 8] 
```
