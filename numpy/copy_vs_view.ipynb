```python
import numpy as np
```

```python
# copy array
```

```python
arr = np.array([10,20,30,40])
copy_arr = arr.copy()
arr[0] = 100

print(arr)
print(copy_arr)
```

```python
# view array
```

```python
arr = np.array([10,20,30,40,50])
view_arr = arr.view()
arr[0] = 100

print(arr)
print(view_arr)
```

```python
# Make Changes in the VIEW
```

```python
arr = np.array([10,20,30])
view_arr = arr.view()
view_arr[0] = 200
print(arr)
print(view_arr)
```
